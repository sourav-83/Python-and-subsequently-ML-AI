```python
import pandas as pd
```

```python
df = pd.read_csv('/kaggle/input/datasets/sourav2083/class-dataset-movies/movie_prediction_dataset_final.csv')
```

```python
df['genre2'] = df['genre2'].fillna('Unknown')
print(df.isnull().sum())
```

```python
most_occured_distributor = df['Distributor'].mode()
print(most_occured_distributor)
```

```python
df['Distributor'] = df['Distributor'].fillna('Warner Bros')
```
